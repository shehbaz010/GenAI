```python
text = """Natural Language Processing is a field of Artificial Intelligence.
It helps computers understand human language.
Dr. Suhashini Chaurasia lives in India and working at RBU."""
print("\nOrignal Text:\n",text)
```

# Step 1: Sentence Segmentation

```python
from nltk.tokenize import sent_tokenize

sentences = sent_tokenize(text)
print("\nSentence Segmentation:")
for i, sent in enumerate(sentences, 1):
    print(f"{i}: {sent}")
```

# Step 2: Word Tokenization

```python
from nltk.tokenize import word_tokenize

words = word_tokenize(text)
print("\nWord Tokenization:")
print(words)
```

# Step 3: Stop Word Removal

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word.lower() not in stop_words]
print("\nAfter Stop Word Removal:")
print(filtered_words)
```

# Step 4: Stemming

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_words]
print("\nAfter Stemming:")
print(stemmed_words)
```

# Step 5: Lemmatization

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]
print("\nAfter Lemmatization:")
print(lemmatized_words)
```

# Step 6: Part-of-Speech (POS) Tagging

```python
pos_tags = nltk.pos_tag(words)
print("\nPOS Tagging:")
print(pos_tags)
```

# Step 7: Named Entity Recognition (NER)

```python
doc = nlp(text)
print("\nNamed Entity Recognition (NER):")
for ent in doc.ents:
    print(f"{ent.text} -> {ent.label_}")
```

# Step 8: Dependency Parsing

```python
print("\nDependency Parsing:")
for token in doc:
    print(f"{token.text} -> {token.dep_} -> {token.head.text}")
```

# Step 9: Noun Phrase Chunking

```python
print("\nNoun Phrase Chunking:")
for chunk in doc.noun_chunks:
    print(chunk.text)
```

```python
%%writefile app.py
import streamlit as st
import nltk
import spacy
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Download NLTK data
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')

# Load Spacy model
try:
    nlp = spacy.load("en_core_web_sm")
except:
    import os
    os.system("python -m spacy download en_core_web_sm")
    nlp = spacy.load("en_core_web_sm")

st.title("NLP Pipeline Showcase")
st.markdown("Analyze text using various Natural Language Processing techniques.")

text_input = st.text_area("Enter your text here:", "Natural Language Processing is a field of Artificial Intelligence. Dr. Suhashini Chaurasia lives in India and working at RBU.")

if st.button("Process Text"):
    # 1. Tokenization
    st.header("1. Tokenization")
    sentences = sent_tokenize(text_input)
    st.write("**Sentences:**", sentences)
    words = word_tokenize(text_input)
    st.write("**Words:**", words)

    # 2. Stopwords & Stemming/Lemmatization
    st.header("2. Text Cleaning")
    stop_words = set(stopwords.words('english'))
    filtered = [w for w in words if w.lower() not in stop_words]

    stemmer = PorterStemmer()
    lemmatizer = WordNetLemmatizer()

    col1, col2 = st.columns(2)
    with col1:
        st.subheader("Stemming")
        st.write([stemmer.stem(w) for w in filtered])
    with col2:
        st.subheader("Lemmatization")
        st.write([lemmatizer.lemmatize(w) for w in filtered])

    # 3. Spacy Features (NER & Parsing)
    st.header("3. Advanced Analysis (Spacy)")
    doc = nlp(text_input)

    st.subheader("Named Entities")
    entities = [(ent.text, ent.label_) for ent in doc.ents]
    if entities:
        st.table(entities)
    else:
        st.write("No entities found.")

    st.subheader("Noun Chunks")
    st.write([chunk.text for chunk in doc.noun_chunks])
```
